### Preprocessing Code

```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import string

# Initialize the tools
lemmatizer = WordNetLemmatizer() #base for changer
stop_words = set(stopwords.words('english')) #stop words removal

def preprocess_text(text):
    # 1. Convert to lowercase
    text = text.lower()
    # print(text)
    
    # 2. Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    # print(text)
    
    # 3. Tokenize (split into words)
    tokens = nltk.word_tokenize(text)
    print(tokens)
    
    # 4. Remove stopwords and Lemmatize
    clean_tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    print(clean_tokens)
    
    return " ".join(clean_tokens)

# Test it out!
sample_q = "How can I apply for a scholarship?"
print(f"Original: {sample_q}")
print(f"Cleaned:  {preprocess_text(sample_q)}")
```

## Processing the whole dataset

#### Step 1: Loading and Cleaning the CSV

```python
import pandas as pd

# 1. Load the dataset we created earlier
faq = pd.read_csv('faq_test_dataset.csv')

# 2. Create a new column 'Cleaned_Question' by applying our function
faq['Cleaned_Question'] = faq['Question'].apply(preprocess_text)

# 3. View the results to ensure it looks right
print("--- Processed Dataframe ---")
faq[['Question', 'Cleaned_Question']]
faq
```

#### Step 2: Vectorization (TF-IDF) Mathematical Phase

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(stop_words='english')
cv_matrix = cv.fit_transform(faq['Cleaned_Question'])
cv.get_feature_names_out()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf = TfidfVectorizer()
tf_matrix = tf.fit_transform(faq['Cleaned_Question'])
# tf_matrix.toarray()
print(tf_matrix.toarray()[:1])
print(tf.get_feature_names_out())
print(tf_matrix.shape)
```

## Step 3: The "Chat" Logic (Cosine Similarity)

```python
from sklearn.metrics.pairwise import cosine_similarity

def get_chatbot_response(user_input):
    # 1. Preprocess the user's question
    cleaned_input = preprocess_text(user_input)
    
    # 2. Transform input to vector (using the SAME vectorizer)
    user_vector = tf.transform([cleaned_input])
    
    # 3. Calculate similarity between user input and all FAQ questions
    # This returns an array of scores between 0 and 1
    similarities = cosine_similarity(user_vector, tf_matrix)
    print("similarity",similarities)
    # 4. Find the index of the highest score
    best_match_index = similarities.argmax()
    max_similarity = similarities[0][best_match_index]
    
    # 5. Set a threshold (if score is too low, the bot didn't understand)
    if max_similarity > 0.2:
        return faq['Answer'].iloc[best_match_index]
    else:
        return "I'm sorry, I don't understand that question. Could you try rephrasing?"

# --- TEST THE CHATBOT ---
print("\n--- Chatbot Test ---")
user_q = "i want to remove my account?"
print(f"User Asked: {user_q}")
```

```python
answer = get_chatbot_response(user_q)
print(answer)
```

```python
print("\n--- FAQ Chatbot is Online (Type 'quit' to exit) ---")

while True:
    user_input = input("You: ")
    
    if user_input.lower() in ['quit', 'exit', 'bye']:
        print("Bot: Goodbye! Have a great day.")
        break
        
    # Process input
    cleaned_input = preprocess_text(user_input)
    user_vector = tf.transform([cleaned_input])
    
    # Calculate Similarity
    similarities = cosine_similarity(user_vector, tf_matrix)
    best_match_index = similarities.argmax()
    confidence = similarities[0][best_match_index]
    
    # Logic for responses
    if confidence > 0.3:  # Adjusted threshold for better precision
        answer = faq['Answer'].iloc[best_match_index]
        print(f"Bot: {answer} (Confidence: {confidence:.2f})")
    else:
        print(f"Bot: I'm not quite sure about that. Can you try asking about shipping, returns, or payments? (Confidence: {confidence:.2f})")
```
